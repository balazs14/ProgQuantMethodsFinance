## Follow-along 1: Built-in numeric types
Python provides `int`, `float`, and `complex` numeric types. Assign one value of each type and inspect them with `type()`.

```python
%reset -f
a = 1  # int
b = 3.14  # float
c = 1 + 1j  # complex

print(type(a))
print(type(b))
print(type(c))
```

## Follow-along 2: Large integers
Integers can be very large. Use exponentiation to create and display a large integer.

```python
%reset -f
long_number = 6 * 10 ** 23 + 1
print(long_number)
```

## Follow-along 3: Even larger integers
Python integers are not limited to ordinary machine-sized values, so the same expression can use a much larger exponent.

```python
%reset -f
very_long_number = 6 * 10 ** 230 + 1
print(very_long_number)
```

## Follow-along 4: Floating-point literals
A decimal point or scientific notation creates a `float`, even when the scientific-notation value is mathematically a whole number.

```python
%reset -f
f1 = 1.0
f2 = 6e23
f3 = 6E23

print(type(f1))
print(type(f2))
print(type(f3))
```

## Follow-along 5: Floating-point representation
Floating-point values are stored in binary, so formatting many decimal places can reveal representation error.

```python
%reset -f
x = 1 / 10
print(x)
s = format(x, ".30f")
print(s)
```

## Follow-along 6: Numeric operation result types
Addition and subtraction combine numeric types, while ordinary division produces a `float`.

```python
%reset -f
a = 1
b = 2
c = 3.0

print(type(a + b))
print(type(a / b))
print(type(a - c))
```

## Follow-along 7: Exponentiation and modulo
The `**` operator calculates powers and `%` returns the remainder after division.

```python
%reset -f
print(3 ** 2)
print(10 % 3)
```

## Follow-along 8: Converting text to numeric types
Use `float()` and `int()` to convert a string to a floating-point number and then to an integer.

```python
%reset -f
s = "3.14"
f = float(s)  # converting text to a floating-point number
i = int(f)  # converting a floating-point number to an integer

print(type(s), s)
print(type(f), f)
print(type(i), i)
```

## Follow-along 9: A conversion that fails
A decimal string cannot be passed directly to `int()`, so this example demonstrates the resulting `ValueError`.

```python
%reset -f
s = "3.14"
i = int(s)
```

## Follow-along 10: Converting numbers to text
`str()` converts numbers to text. Numeric addition and string concatenation then produce different results.

```python
%reset -f
i1 = 5
i2 = 10
s1 = str(i1)
s2 = str(i2)

print(i1 + i2)
print(s1 + s2)
```

## Follow-along 11: Boolean values and comparisons
Boolean values are `True` or `False`. Comparison operators such as `==`, `!=`, `<`, and `>=` produce boolean results.

```python
%reset -f
a, b = 10, 15

x = a > b
y = a > 0 and a < 100
z = x or y

print(x)
print(y)
print(z)
```

## Follow-along 12: Reading text input
`input()` displays a prompt, waits for the user to press Enter, and stores the entered text in a variable.

```python
%reset -f
message = input("Say something and I will repeat it: ")
print(message)
```

## Follow-along 13: Input is text by default
Even numeric-looking input is initially a string, so attempting arithmetic before conversion raises a `TypeError`.

```python
%reset -f
x = input("Give me a number and I will tell you its square: ")
print(x ** 2)
```

## Follow-along 14: Converting input before arithmetic
Convert the input to `int` before calculating its square.

```python
%reset -f
x = input("Give me a number and I will tell you its square: ")
x = int(x)
print(x ** 2)
```

## Follow-along 15: A basic `if` statement
The indented block runs only when the condition after `if` is true.

```python
%reset -f
a = 7
b = 13
if b > a:
    print("b is greater than a")
```

## Follow-along 16: The one-line form of `if`
When the branch has only one instruction, Python also permits the instruction on the same line.

```python
%reset -f
a, b = 7, 13  # this can be shortened here too
if b > a: print("b is greater than a")
```

## Follow-along 17: The `else` branch
Use `else` to run a different indented block when the `if` condition is false.

```python
%reset -f
a, b = 17, 13
if b > a:
    print("b is greater than a")
else:
    print("b is less than or equal to a")
```

## Follow-along 18: Nested conditions
A second `if` can be nested inside the `else` branch to handle a third possibility.

```python
%reset -f
a, b = 13, 13
if b > a:
    print("b is greater than a")
else:
    if b == a:
        print("b is equal to a")
    else:
        print("b is less than a")
```

## Follow-along 19: The `elif` form
`elif` expresses the middle branch more compactly than nesting a second `if` inside `else`.

```python
%reset -f
a, b = 13, 13
if b > a:
    print("b is greater than a")
elif b == a:
    print("b is equal to a")
else:
    print("b is less than a")
```

## Follow-along 20: The `pass` statement
Use `pass` as a temporary placeholder when a conditional branch is intentionally empty during development.

```python
%reset -f
a, b = 17, 13
if b > a:
    pass
else:
    print("b is less than or equal to a")
```

## Follow-along 21: Checking converted input
Convert valid numeric input and calculate its square. This version does not yet handle invalid input.

```python
%reset -f
x = input("Give me a number and I will tell you its square: ")
x = int(x)
print(x ** 2)
```

## Follow-along 22: Validating numeric input
`isnumeric()` checks whether the input text contains a numeric value before converting it. The invalid-input message uses `\n` for a line break.

```python
%reset -f
x = input("Give me a number and I will tell you its square: ")
if x.isnumeric():
    x = int(x)
    print(x ** 2)
else:
    print("Invalid input!\nOnly a numeric value may be given.")
```
