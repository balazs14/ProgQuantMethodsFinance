### Quiz 1 — Exercise 2: Locating a country code in an ISIN

A bond's ISIN is stored as:

```
isin = "HU0000403706"
```

Write code that:
1. Checks whether `"HU"` appears in `isin`.
2. Finds the starting position of `"HU"` within `isin`.
3. Uses slicing (based on the position found, not a hard-coded index) to extract the two-letter country code.

Print all three results, one per line.

**This cell must run entirely on its own, with a single "Run Cell" command.** Include any imports it needs directly in this cell (none are required for this exercise).

````python
isin = "HU0000403706"
contains_hu = "HU" in isin
position = isin.find("HU")
country_code = isin[position:position + 2]
print(contains_hu)
print(position)
print(country_code)
````
