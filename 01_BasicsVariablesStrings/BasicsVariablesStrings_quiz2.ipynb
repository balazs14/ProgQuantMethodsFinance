### Quiz 2 — Exercise 1: Cleaning up a fund name

An analyst pasted a messy fund name into a variable:

```
raw_name = "  danube VALUE fund  "
```

Write code that, using as few lines as possible:
1. Removes the leading and trailing spaces.
2. Prints the cleaned fund name in Title Case.
3. Prints the length of the cleaned (stripped) name.

**This cell must run entirely on its own, with a single "Run Cell" command.** Include any imports it needs directly in this cell (none are required for this exercise).

````python
raw_name = "  danube VALUE fund  "
clean_name = raw_name.strip()
print(clean_name.title())
print(len(clean_name))
````

### Quiz 2 — Exercise 2: Formatting a bond coupon

A bond is described by:

```
bond_name = "Danube Treasury 2031"
coupon_rate = 0.0453
```

`coupon_rate` is stored as a decimal (i.e. 0.0453 means 4.53%). Using an f-string, print a single line reading exactly:

```
Bond: Danube Treasury 2031, Coupon: 4.53%
```

with the coupon shown as a percentage to two decimal places.

**This cell must run entirely on its own, with a single "Run Cell" command.** Include any imports it needs directly in this cell (none are required for this exercise).

````python
bond_name = "Danube Treasury 2031"
coupon_rate = 0.0453
print(f"Bond: {bond_name}, Coupon: {coupon_rate * 100:.2f}%")
````
