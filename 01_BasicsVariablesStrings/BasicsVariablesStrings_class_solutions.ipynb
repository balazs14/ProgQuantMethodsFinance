## Exercise 10
Remove leading and trailing spaces, then print the result length

```python
%reset -f
label = "   Danube Value Fund   "
print(len(label.strip()))

# Without using the dedicated function
left = 0
right = len(label)
while left < right and label[left] == " ":
    left += 1
while right > left and label[right - 1] == " ":
    right -= 1
print(right - left)
```
