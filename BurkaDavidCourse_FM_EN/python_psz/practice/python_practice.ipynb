Írj egy függvényt, ami paraméterként megkapja egyháromszög három oldalának hosszát, és kiszámítja egy háromszög területét (__[Hérón képlet](https://hu.wikipedia.org/wiki/H%C3%A9r%C3%B3n-k%C3%A9plet)__) és a szögeit (__[Koszinusztétel](https://hu.wikipedia.org/wiki/Koszinuszt%C3%A9tel)__), amiket egy listában ad vissza. Használd a math.sqrt, math.acos és math.degrees függvényeket.

```python
import math

def haromszog_terulet_es_szogek(a, b, c):
    s = (a + b + c) / 2
    terulet = math.sqrt(s * (s - a) * (s - b) * (s - c))
    szogA = math.degrees(math.acos((b*b + c*c - a*a) / (2 * b * c)))
    szogB = math.degrees(math.acos((a*a + c*c - b*b) / (2 * a * c)))
    szogC = 180 - szogA - szogB
    return [terulet, szogA, szogB, szogC]

haromszog_terulet_es_szogek(5,12,13)
```

3. Írj egy függvényt, ami kiszámítja egy kör és az azt érintő téglalap területének arányát, ha meg van adva a kör sugara. Használd a math.pi és math.pow függvényeket.

```python
import math

def terulet_arany(r):
    kor_terulete = math.pi * math.pow(r, 2)
    teglalap_terulete = 2 * r * 2 * r
    arany = kor_terulete / teglalap_terulete
    return arany

terulet_arany(2)
```

6. Trigonometriai szögfüggvények gyakorlati alkalmazása
    1.	Importáld be a Math modult.
    2.	Kérd be a felhasználótól egy háromszög egyik szögét fokban (gamma) és a mellette lévő két oldal (a és b) hosszát. Alakítsd ezeket az értékeket lebegőpontos számokká.
    3.	Alakítsd át a szög értékét radiánra.
    4.	Számítsd ki a harmadik oldal (c) hosszát a koszinusz-tétel segítségével.
    5.	Számítsd ki a háromszög területét. 
    6.	Számítsd ki a szöghöz (gamma) tartozó magasságot.
    7.	Írasd ki az eredményeket két tizedesjegy pontossággal!
    
    Képletek:

    - Koszinusz-tétel: c2 = a2 + b2 – 2*a*b*cosγ 
    - Terület: T = a*b*sinγ
    - Magasság: mc = 2T / c

```python
import math

# Bemenetek a felhasználótól:
gamma = int(input("Adja meg a háromszög gamma szögét (fokban): "))
a = int(input("Adja meg az 'a' oldal hosszát: "))
b = int(input("Adja meg a 'b' oldal hosszát: "))

# Gamma szög átváltása radiánra:
gamma_rad = math.radians(gamma)

# Koszinusz-tétel: harmadik oldal kiszámítása:
c = math.sqrt(a**2 + b**2 - 2 * a * b * math.cos(gamma_rad))

# Terület kiszámítása:
terulet = 0.5 * a * b * math.sin(gamma_rad)

# Magasság kiszámítása:
gamma_magassag = (2 * terulet) / c

# Kiíratás:
print(f"A háromszög 'c' oldala: {c:.2f}")
print(f"A háromszög területe: {terulet:.2f}")
print(f"A gamma szöghöz tartozó magasság: {gamma_magassag:.2f}")
```

### datetime

1. Írj egy függvényt, aminek két egész szám paramétere van: év és hónap. A függvény visszatérési értéke legyen egy lista, amely a megadott év és hónap összes napját tartalmazza. Használd a datetime modul date és timedelta osztályait.

```python
from datetime import date, timedelta

def honap_napjai(ev, honap):
    napok = []
    current_date = date(ev, honap, 1)
    while current_date.month == honap:
        napok.append(current_date.strftime('%Y-%m-%d'))
        current_date += timedelta(days=1)
    return napok

honap_napjai(2024,11)
```

```python
from datetime import datetime, time

def munkaido_e(ido):
    munkaido_start = time(9, 0)
    munkaido_end = time(17, 0)
    return munkaido_start <= ido.time() <= munkaido_end

munkaido_e(datetime.now())
```

```python
from datetime import datetime, timedelta

def jovobeli_datum(napok):
    today = datetime.today()
    future_date = today + timedelta(days=napok)
    return future_date.strftime("%Y-%m-%d")

jovobeli_datum(15)
```

5. Írj egy függvényt, aminek két egész szám paramétere van: év és hónap. A függvény kiszámítja az adott hónapban található szabadnapok (hétvégék) számát. Használd a datetime modul date és timedelta osztályait illetve a weekday függvényt.

```python
from datetime import date, timedelta

def szabadnapok_szama(ev, honap):
    szabadnapok = 0
    current_date = date(ev, honap, 1)
    while current_date.month == honap:
        if current_date.weekday() >= 5:  # Szombat vagy vasárnap
            szabadnapok += 1
        current_date += timedelta(days=1)
    return szabadnapok

szabadnapok_szama(2024,11)
```

### pandas

1. Írj egy függvényt, ami paraméterként kap egy pandas DataFrame-et, és kezeli a hiányzó értékeket a következő módon:
* a nem numerikus (dtype=='object') oszlopokat pedig a leggyakoribb értékkel
* a numerikus oszlopokat az oszlop átlagával.

Ezután olvasd be az *olimpia.csv* fájlt, és hívd meg vele a függvényt. Vizsgáld meg, hogy valóban feltöltötte-e az üres értékeket.

```python
import pandas as pd

def tisztit_adat(df):
    for column in df.columns:
        print(df[column].dtype)
        if df[column].dtype == 'object':  # Ha az oszlop nem numerikus
            df[column] = df[column].fillna(df[column].mode()[0])
        else:  # Ha az oszlop numerikus
            df[column] = df[column].fillna(df[column].mean())
    return df

df = pd.read_csv('olimpia.csv')

print(df.info())

print(tisztit_adat(df).info())
```
