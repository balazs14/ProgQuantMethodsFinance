```python
# 1. Feladat

import pandas as pd
df = pd.read_csv('games.csv', index_col=0)

df['Date'] = pd.to_datetime(df['Year'], format='%Y')

df['EU_Sales_%'] = (100 * df['EU_Sales'] / df['Global_Sales']).round(2)

df[df['Date'].dt.is_leap_year].loc[:50][['Name', 'Publisher', 'EU_Sales_%']]
```

```python
# 2. Feladat

import pandas as pd
import plotly.express as px

df = pd.read_csv('games.csv', index_col=0)

fig = px.scatter_3d(
    df, 
    x = 'NA_Sales', 
    y = 'EU_Sales', 
    z = 'JP_Sales', 
    color='Publisher', 
    hover_name='Name')
fig.show()

dfg = df.sort_values('Global_Sales').groupby(['Platform', 'Year'])

dfn = pd.DataFrame()
dfn['Best_game'] = dfg.last()['Name']
dfn['Total_sales'] = dfg['Global_Sales'].sum()
dfn.reset_index(inplace=True)

display(dfn)

fig2 = px.bar(
    dfn[dfn['Year']==2010], 
    x = 'Platform', 
    y = 'Total_sales',
    hover_name='Best_game')
fig2.show()
```
